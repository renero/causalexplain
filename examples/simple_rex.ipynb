# Simple Usage Example of CausalExplain

This is a simple example of how to use the `causalexplain` package to perform causal graph discovery using the REX model on a toy dataset with four variables.


## Notes

**Note 1**: 

> If you cannot import `causalexplain`, run the notebook from the repo root, or install the package (`pip install -e .`), or add the repo root to `sys.path`. 

**Note 2**: 

> Warning messages from the `tqdm` package can be solved by installing or updating IPwidgets: `pip install -U ipywidgets jupyter`.

**Note 3**: 

> For higher-quality math text in plots, install a LaTeX distribution; otherwise pass `usetex=False` when plotting.

To train and evaluate a causal discovery model, create a `GraphDiscovery` object with the appropriate parameters, and then call the `run()` method to execute the experiments. Finally, visualize the results using the `plot()` method.

```python
from causalexplain import GraphDiscovery

experiment = GraphDiscovery(
    experiment_name='my_experiment',
    model_type='rex',
    csv_filename='../data/toy_dataset.csv',
    true_dag_filename='../data/toy_dataset.dot')

# Run the experiments
experiment.run(hpo_iterations=2, bootstrap_samples=2, combine_op='union', quiet=True)

# Local runs without LaTeX: disable usetex to avoid texmanager errors.
experiment.plot(show_metrics=True, layout='circular', usetex=False)
```

If you want to save the trained model for future use, you can use the `save()` method. To load a previously saved model, create a `GraphDiscovery` object and call the `load()` method with the filename of the saved model.

```python
experiment.save('simple_rex_experiment')
```

## Loading a pretrained model

If you prefer to load a pretrained model, then simply create the object from `GraphDiscovery()` and call the `load()` and `plot()` methods.

```python
from causalexplain import GraphDiscovery

experiment = GraphDiscovery()
experiment.load('simple_rex_experiment.pickle')
experiment.plot(show_metrics=True, layout='circular', usetex=False)
```
